# Hasil dan Pengujian Hipotesis

Bandingkan A-D dengan statistik deskriptif. Uji signifikansi antarfold bersifat eksploratif jika metadata subjek/KSS masih dummy.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
summary = pd.read_csv('../results/summary.csv')
display(summary)
plot = summary.melt(id_vars='scenario', value_vars=['accuracy_mean', 'f1_weighted_mean'], var_name='metric', value_name='score')
sns.barplot(data=plot, x='scenario', y='score', hue='metric')
plt.ylim(0, 1)
plt.show()
```
