# Evaluasi Cross-Validation

Buka metrik dan confusion matrix dari lima fold tiap skenario. Validasi berbasis pseudo-group/dummy bukan bukti generalisasi antar-subjek.

```python
from pathlib import Path
import pandas as pd
for scenario in 'ABCD':
    metrics = pd.read_csv(f'../results/{scenario}.csv')
    print(f'\nScenario {scenario}')
    display(metrics)
    matrices = sorted(Path('../results').glob(f'{scenario}_fold*_cm.csv'))
    print('Confusion matrices:', [path.name for path in matrices])
```
